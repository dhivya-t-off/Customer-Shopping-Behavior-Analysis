```python
import pandas as pd
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns = df.columns.str.replace(" ","_")
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
lables = ['Young Adult','Adult','Middle Aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=lables)
```

```python
df.columns
```

```python
df[['age','age_group']].head()
```

```python
df.isnull().sum()
```

```python
df['review_rating'] = df.groupby('category')['review_rating'].transform(lambda x:x.fillna(x.mean()))
```

```python
df.isnull().sum()
```

```python
df['frequency_of_purchases'].unique()
```

```python
frequency_map = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Every 3 Months': 90,
    'Quarterly': 90,
    'Annually': 365
}
df['frequency_purchases_days']=df['frequency_of_purchases'].map(frequency_map)
```

```python
df[['frequency_purchases_days','frequency_of_purchases']].head()
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used',axis=1)
```
